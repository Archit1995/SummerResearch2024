```python
# Finite field F = Z/107Z
p = 107
F = GF(p)

# Example: polynomial ring in n variables (X1,...,Xn)
# You can adjust n and names as needed.
n = 2  # for example; change to your desired number of variables
R = PolynomialRing(F, n, 'x')
R
```

```python
K = Frac(R)
x=R.gens()
x
```

```python
f=x[0]+5*x[1]
g=x[0]*x[1]+3
h=K(f)/K(g)
h
```

```python
def restricting_rational_function_to_line(h, betas, sigmas, alpha, p=107, return_int=True):
    """
    Restrict a 2-variable rational function h(X,Y) over GF(p) to the line
    psi(x) = (x, β2*(x - σ1) + σ2), then evaluate at x = alpha.

    betas:  [β2]      (list of length 1)
    sigmas: [σ1, σ2]  (list of length 2)
    alpha:  element of GF(p) or integer

    Returns h(psi(alpha)) in GF(p) or as int if return_int=True.
    """

    F = GF(p)
    sigmas = [F(s) for s in sigmas]
    betas  = [F(b) for b in betas]
    alpha  = F(alpha)

    if len(sigmas) != 2:
        raise ValueError("For n=2, sigmas must be [σ1, σ2].")
    if len(betas) != 1:
        raise ValueError("For n=2, betas must be [β2].")

    sigma1, sigma2 = sigmas
    beta2 = betas[0]

    # h is in Frac(R) where R = F[X,Y]
    num = h.numerator()
    den = h.denominator()
    R = num.parent()
    X, Y = R.gens()

    # Polynomial ring in t
    S.<t> = PolynomialRing(F)

    # Images under psi: X -> t, Y -> β2*(t-σ1) + σ2
    X_image = t
    Y_image = beta2*(t - sigma1) + sigma2

    phi = R.hom([X_image, Y_image], S)

    num_t = phi(num)
    den_t = phi(den)

    num_val = num_t(alpha)
    den_val = den_t(alpha)

    if den_val == 0:
        # Pole: h(psi(alpha)) is undefined
        return None

    value = num_val / den_val   # in F

    if return_int:
        return int(value)
    else:
        return value
```

```python
sigma=[2,3];
beta=[7];
alphas=[1,2,3,4,5];
results = {}  # dictionary: i -> list of (alpha, value)
for i in range(8):
    sigma_i = [F(2)^i, F(3)^i]  # sigma^(i) = [2^i, 3^i] in GF(107)
    line_values = []
    for a in alphas:
        val = restricting_rational_function_to_line(h, beta, sigma_i, a, p=p, return_int=True)
        line_values.append((a, val))
    results[i] = {
        "sigma": sigma_i,
        "values": line_values
    }

results
```

```python
for i in range(8):
    print(f"Line L_{i} with sigma^{i} = {results[i]['sigma']}")
    for a, v in results[i]['values']:
        print(f"  x = {a} -> h(psi_{i}({a})) = {v}")
    print()
```

```python
from sage.plot.colors import rainbow

def plot_sampled_lines_and_fit(h, p=107):
    """
    Plot only the sampled points (alpha_j, L_i(alpha_j)) for
    j = 1..5 and i = 0..7, color-coded by i.

    Reuses the existing restricting_rational_function_to_line(h, betas, sigmas, alpha).
    No line fitting is done anymore.
    """
    F = GF(p)
    betas = [7]                    # fixed direction
    alphas = [1, 2, 3, 4, 5]       # chosen x-values on each line
    colors = rainbow(8)

    G = Graphics()
    results = {}

    for i in range(8):
        # sigma^(i) = [2^i, 3^i] in GF(p)
        sigmas_i = [F(2)^i, F(3)^i]

        vals = []

        for a in alphas:
            aF = F(a)

            # Compute the actual point on the line:
            # psi_i(a) = (x, y) with y = 7*(a - sigma1) + sigma2
            x_val = aF
            y_val = betas[0] * (aF - sigmas_i[0]) + sigmas_i[1]

            x_int = int(x_val)
            y_int = int(y_val)

            # Plot the point
            G += point((x_int, y_int), size=40, color=colors[i])

            # Label the point with its coordinates
            G += text(f"({x_int},{y_int})",
                      (x_int + 0.1, y_int + 0.8),
                      color=colors[i], fontsize=6)

            # Evaluate the restricted rational function at this alpha
            val = restricting_rational_function_to_line(
                h, betas, sigmas_i, a, p=p, return_int=True
            )
            vals.append((a, val))

        results[i] = {"sigma": sigmas_i, "values": vals}

    G.show(figsize=(7, 7))
    return results
```

```python
vals = plot_sampled_lines_and_fit(h, p=107)
vals
```
